each step ⬇️

```python
import re


# Create a class for our simple tokenizer
class SimpleTokenizerV1:

    # __init__ runs automatically when we create the tokenizer object
    def __init__(self, vocab):

        # Store the vocabulary dictionary
        # token -> integer
        # Example: {"Hello": 0, ",": 1, "world": 2}
        self.str_to_int = vocab

        # Create the reverse vocabulary
        # integer -> token
        # vocab.items() gives:
        # token = s
        # integer = i
        #
        # Example:
        # {"Hello": 0, ",": 1, "world": 2}
        #
        # becomes:
        # {0: "Hello", 1: ",", 2: "world"}
        self.int_to_str = {i: s for s, i in vocab.items()}


    # This method converts text into integer IDs
    # Text -> Tokens -> Integer IDs
    def encode(self, text):

        # Split the text into tokens
        #
        # [,.:;?_!"()']  -> punctuation marks
        # |              -> OR
        # --             -> treat double hyphen as a separate token
        # \s             -> whitespace such as space, tab, newline
        #
        # The () around the whole pattern are used to KEEP
        # the matched punctuation/whitespace as separate items.
        preprocessed = re.split(
            r'([,.:;?_!"()\']|--|\s)',
            text
        )

        # Remove unnecessary whitespace from each item
        # and remove empty strings
        #
        # item.strip() -> removes spaces from beginning/end
        # if item.strip() -> keeps only non-empty items
        preprocessed = [
            item.strip()
            for item in preprocessed
            if item.strip()
        ]

        # Convert every token into its integer ID
        #
        # self.str_to_int stores:
        # token -> integer
        #
        # Example:
        # "Hello" -> 0
        # ","     -> 1
        # "world" -> 2
        ids = [
            self.str_to_int[s]
            for s in preprocessed
        ]

        # Return the list of integer IDs
        return ids


    # This method converts integer IDs back into text
    # Integer IDs -> Tokens -> Text
    def decode(self, ids):

        # Convert every integer ID back into its token
        #
        # self.int_to_str stores:
        # integer -> token
        #
        # Example:
        # 0 -> "Hello"
        # 1 -> ","
        # 2 -> "world"
        tokens = [
            self.int_to_str[i]
            for i in ids
        ]

        # Join all tokens together with a space
        #
        # Example:
        # ["Hello", ",", "world", "!"]
        #
        # becomes:
        # "Hello , world !"
        text = " ".join(tokens)

        # Remove unnecessary spaces before punctuation
        #
        # \s+ -> one or more whitespace characters
        # (...) -> capture the punctuation
        # \1 -> put the captured punctuation back
        #
        # Example:
        # "Hello , world !"
        #
        # becomes:
        # "Hello, world!"
        text = re.sub(
            r'\s+([, .?! "()\'])',
            r'\1',
            text
        )

        # Return the final decoded text
        return text
```

**SimpleTokenizerV2**

```python
class SimpleTokenizerV2:
  def __init__(self,vocab):
    self.str_to_int = vocab #store the vocab
    self.int_to_str = {i:s for s,i in vocab.items()} # this is dictornary {} i = key s= value

  def encode(self,text):
    preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)',text)
    preprocessed = [item.strip() for item in preprocessed if item.strip()]#item.strip()removes whitespace from the beginning and end.
    preprocessed = [item if item in self.str_to_int #[value_if_true if condition else value_if_false for item in list]
                    else "<|unk|>" for item in preprocessed] #In one sentence: this line replaces every token that is not present in the vocabulary with <|unk|>.

    ids =  [self.str_to_int[s] for s in preprocessed]
    return ids

  def decode(self, ids):
    text = " ".join( [self.int_to_str[i] for i in ids])
    # Replace spaces before the specified punctuations
    text = re.sub(r'\s+([, .?! "()\'])', r'\1', text)
    return text
```

here u can see that now it handle the unknown word also and seperated the 2 sources with the endoftext

```python
tokenizer = SimpleTokenizerV2(vocab)

text1 = "Hello, do you like tea?"
text2 = "In the sunlit terraces of the palace."

text ="<|endoftext|> ".join((text1, text2) )

print(text)
```
