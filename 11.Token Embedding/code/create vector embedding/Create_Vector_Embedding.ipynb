```python
import torch
```

# Create input token IDs

**torch.tensor([...]) converts the list of integers into a PyTorch tensor. The values [2, 3, 5, 1] are token IDs, not the words themselves. Each ID must be a valid row index in the embedding table.
Because the table will have six rows, valid IDs are 0, 1, 2, 3, 4, 5**

```python
# Store four token IDs in a tensor. These IDs will be used to look up four vectors later.
input_ids = torch.tensor([2, 3, 5, 1])

# Show the tensor so you can confirm the IDs being passed to the embedding layer.
print(input_ids)
```

#Choose the embedding table's dimensions
- vocab_size = 6: the table has six rows, one for each possible token ID.

- output_dim = 3: each token is represented by a vector with three numbers.

**Therefore, the weight matrix will have shape (6, 3): six rows and three columns. The original notebook uses the name vocal_size; vocab_size is a clearer, conventional spelling.**

```python
# Number of token IDs the embedding table can look up (IDs 0 through 5).
vocab_size = 6

# Number of values in each token's embedding vector.
output_dim = 3
```

# Set the random seed and create the embedding layer

**torch.manual_seed(123)** sets PyTorch's random seed so that the initial random weights can be reproduced in the same environment.

**torch.nn.Embedding(vocab_size, output_dim)** creates a trainable lookup table. It initially fills the table with values that are generally random. It does not automatically learn meaning at creation time; training must update these weights.**

```python
# Set a seed so the initial random weights are repeatable in the same setup.
torch.manual_seed(123)

# Create an embedding table with 6 rows and 3 values in each row.
embedding_layer = torch.nn.Embedding(vocab_size, output_dim)

# The layer's weight matrix has shape (6, 3).
print('Weight matrix shape:', embedding_layer.weight.shape)
```

#Inspect the embedding weight matrix
**embedding_layer.weight** accesses the trainable matrix stored inside the layer. Each row belongs to one token ID:
- row 0 → token ID 0's vector
- row 1 → token ID 1's vector
- row 2 → token ID 2's vector, and so on
The row number is the token ID, but remember that counting starts at zero. The three columns are the three dimensions of each vector.

```python
# Print all six embedding vectors (the complete 6-by-3 weight matrix).
# These are initial values, not trained semantic representations yet.
print(embedding_layer.weight)
```

# Look up the vector for token ID 3

**torch.tensor([3])** creates a tensor containing one token ID. Passing it to embedding_layer(...) asks the layer to return the vector from row 3.

Because IDs start at zero, token ID 3 is stored in the fourth row. The output shape is (1, 3): one requested token and three embedding values.

```python
# print(embedding_layer(torch.tensor([3])))

                #or

# Make a one-item tensor containing token ID 3.
single_id = torch.tensor([3])

# Look up and print the embedding vector for token ID 3.
vector_for_id_3 = embedding_layer(single_id)
print(vector_for_id_3)

# Confirm the output shape: 1 token by 3 embedding dimensions.
print('Output shape:', vector_for_id_3.shape)
```

# Look up all four input IDs at once

**The same layer can look up multiple IDs in one call. For input_ids = [2, 3, 5, 1], it returns rows 2, 3, 5, and 1 in that same order. The output shape is (4, 3) because there are four IDs and each vector has three values.**

```python
# Look up the embedding vector for each ID in input_ids, preserving their order.
input_embeddings = embedding_layer(input_ids)

# Print the four vectors. Each output row corresponds to one input ID.
print(input_embeddings)

# Expected shape: 4 input IDs by 3 values per embedding vector.
print('Output shape:', input_embeddings.shape)
```
