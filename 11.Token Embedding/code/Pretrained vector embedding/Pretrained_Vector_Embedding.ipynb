# Let us check the similarity between a few pairs of words

 Cosine Similarity Score Interpretation:

1.0  -> Exactly the same vector direction

        Example: dog vs dog

0.9  -> Very high similarity

         Example: king vs queen (possible model result)

0.85 -> High similarity

         Example: boy vs girl (actual result: 0.854)

0.5  -> Moderate positive similarity

         Example: dog vs puppy (possible model result)

0.3  -> Relatively low positive similarity

         Example: dog vs girl (actual result: 0.343)

0.0  -> Perpendicular vector directions

         Meaning: No directional alignment

-0.5  -> Negative cosine similarity

         Meaning: Vectors point in substantially different directions

-1.0  -> Opposite vector directions

         Meaning: Vectors point in exactly opposite directions

 Important:

 - Higher cosine similarity means more aligned vector directions.

 - Cosine similarity is NOT a percentage of similarity.

 - Cosine similarity measures direction, not direct distance.

 - Word-pair scores are examples, not guaranteed results.

```python
# Example of calculating similarity
print(word_vector.similarity( 'women', 'man'))
print(word_vector.similarity('king', 'queen'))
print(word_vector.similarity('uncle', 'aunt'))
print(word_vector.similarity('boy', 'girl'))
print(word_vector.similarity('nephew', 'niece'))
print(word_vector. similarity( 'paper', 'water'))
```

**Magnitude:- it tells you how long the difference vector is between two word embeddings.**

- Small magnitude: The vectors' endpoints are close together.
- Large magnitude: The vectors' endpoints are farther apart.

```python
import numpy as np
# Words to compare
word1 = 'man'
word2 = 'woman'

word3 = 'semiconductor'
word4 = 'earthworm'

word5 = 'nephew'
word6 = 'niece'

# Calculate the vector difference
vector_difference1 = model[word1] - model [word2]
vector_difference2 = model [word3] - model [word4]
vector_difference3 = model [word5] - model [word6]

# Calculate the magnitude of the vector difference
magnitude_of_difference1 = np.linalg.norm(vector_difference1) #np.linalg.norm() calculates the magnitude (length) of that difference vector.
magnitude_of_difference2 = np.linalg.norm(vector_difference2)
magnitude_of_difference3 = np.linalg.norm(vector_difference3)
```

```python
# Print the magnitude of the difference
print(r"The magnitude of the difference between '{}' and '{}' is {:.2f}".format(word1, word2, magnitude_of_difference1))
print(r"The magnitude of the difference between '{}' and '{}' is {:.2f}".format(word3, word4, magnitude_of_difference2))
print(r"The magnitude of the difference between '{}' and '{}' is {:.2f}".format(word5, word6, magnitude_of_difference3))
```
