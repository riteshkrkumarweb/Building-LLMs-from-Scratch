```python
import tiktoken
encoding = tiktoken.get_encoding("cl100k_base")
#encoding = tiktoken.encoding_for_model("gpt2")
```

```python
text = ('My name is Ritesh, and I am learning Artificial Intelligence and Machine Learning.'
  'I have been exploring Python, Deep Learning, Natural Language Processing,'
  'and Large Language Models. I enjoy understanding how things work behind the scenes, '
  'especially how computers process human language.'
  'My goal is to build useful AI projects and continuously improve my programming skills.')
```

**Encoding Text**

```python
tokens = encoding.encode(text) # encoding with byte pair encoding
print(tokens)
```

```python
for token in tokens:
  print(token, "-->" ,encoding.decode_single_token_bytes(token)) #b mean byte
```

**Decoding Text**

```python
decode = encoding.decode(tokens)
print(decode)
```
